# 1. Salarios mensuales en quetzales

| Rol                     | Salario mensual |
| ----------------------- | --------------: |
| Junior Mobile Developer |          Q9,360 |
| Senior Mobile Developer |         Q27,300 |
| Scrum Master            |         Q19,500 |
| Project Manager         |         Q23,400 |

---

# 2. Cálculo mensual por rol

### Juniors

$$3 \times 1.0 \times Q9{,}360 = Q28{,}080$$

### Senior

$$0.6 \times Q27{,}300 = Q16{,}380$$

### Scrum Master

$$0.2 \times Q19{,}500 = Q3{,}900$$

### Project Manager

$$0.2 \times Q23{,}400 = Q4{,}680$$

---

# 3. Costo mensual total

$$Q28{,}080 + Q16{,}380 + Q3{,}900 + Q4{,}680 = Q53{,}040$$

## Costo mensual del equipo Scrum
# **Q53,040/mes**

---

# 4. Proyecto de 3 meses (6 sprints)

$$Q53{,}040 \times 3 = Q159{,}120$$

## Costo total del proyecto
# **Q159,120**

---

# 5. Margen del 30%

## Precio al cliente

$$\text{Precio al cliente} = \frac{Q159{,}120}{1 - 0.30} = \frac{Q159{,}120}{0.70} = Q227{,}314.29$$

## Precio final de venta
# **Q227,314**

---

# 6. Ganancia esperada

$$Q227{,}314 - Q159{,}120 = Q68{,}194$$

## Ganancia
# **Q68,194**

---

# 8. Fórmula general en GTQ

## Costo total

$$\text{CostoTotal} = \text{Meses} \times \left[ (3 \times \text{Junior} \times 1.0) + (\text{Senior} \times 0.6) + (\text{ScrumMaster} \times 0.2) + (\text{ProjectManager} \times 0.2) \right]$$

## Precio al cliente

$$\text{PrecioCliente} = \frac{\text{CostoTotal}}{1 - \text{Margen}}$$

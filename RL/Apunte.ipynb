### EL CICLO AGENTE - ENTORNO. 

AGENTE ----(ACCIONES)-----> ENTORNO

EL entorno responde hacia el agente

AGENTE <----(RECOMPENSA | SIGUIENTE ESTADO)----- ENTORNO

Recompensa se define asi $R_{T+1}$ porque sucede en el siguiente estado.

TRAYECTORIA 

$$T: (S_0 , A_0, R1, S_1 , A_1, R2.............S_N , A_N, R_{N+1})$$

S : estado 
A: accion
R: recompensa

#### Propiedad de MARKOV
El estado actual solo depende del estado anterior. La probabilidad de transicionar al estado T+1 dado en el estado que estoy actualmente. 

El agente va a observar el estado , selecciona una A segun su politica (es el cerebro del agente), el entorno transiciona al estado siguiente y emite la recompensa correspondiente.



## ESTADO, OBSERVACION Y ESPACIO DE ACCIONES

- Estado $S_t$ :  Representacion completa del entorno puede ser inaccesible. Si tuviera acceso a todo el estado. Podriamos predecir la accion siguiente.

- Observacion $O_t$: Lo que el agente realmente percibe es parcial 

- Entorno totalmente observable: $O_t = S_t -> MDP$

- Entorno parcialmente observable $O_t != S_t -> POMDP$



## La recompensa
Es el unico canal de comunicación es entre el diseñador y el sistema como tal. La recompensa es que todo objetivo se puede expresar como maximización de recompensa, esperamos que exista una recompensa a futuro. La recompensa es señal de diseño , no dato observado.

La recompensa la decidimos en ese momento. 

### Ejemplos
- Banca: +1 si el credito se paga, -1 si entra en mora.
- TELCO: Recompensa proporcional al throughput del canal. 
- SALUD: Días sin sintomas como proxy de bienestar del paciente. 


Esto tiene desventajas:

- Mal diseño: En una carrera, EL hecho de acelerar nos da puntaje pero no estamos terminando la carrera, Entonces puede entrar en un bucle que acelera pensando que asi gana puntos pero realmente no termina la carrera.

### POLITICA

El aprendizaje por RL busca la mejor politica posible, la politica se denomia con PI y mapea estados a acciones. Esta primera puede ser deterministtica o estocastica

- DETERMINISTICA : Mapea estado a accion
- ESTO CASTICA: Distribución sobre las acciones. 


Cuando es deterministica la politica determina la misma acción. 

La poltica siempre se optimiza para buscar la mejor. 

Evoluciona en el entrenamiento.

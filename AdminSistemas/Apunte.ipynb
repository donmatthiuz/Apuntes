##  ITSM
El modelo operativo que transforma la capacidad tecnica en resultados de negocio. 

Se enfoca en el ciclo de vida completo de los servicios de IT.

Gestiona la disponiblidad , el rendimiento y utilidad. 



### PILARES

Se enfoca en las personas.

- Personas: Toda la estructura, y la cultura de servicio.
- Procesos: Flujo de trabajo estandarizados como gestion de incidentes, cambios o problemas. 
- Tecnologia: Las herramientas como Jira Service Now, que automatizan y miden los prcesos. 


### RELEVANCIA

- Mitigar el caos. Psar de una cultura de apaga fuegos. 
- Alineación estrategica. Asegurar en el presupuesto y el esfruerzo de ingieria estan dedicados a servicios que realmenten e impactan
- Cumplimineto y gobernanza. Facitilitar auditorias y garantizar que los niveles de servicios se cumplan de manera sistematica. 0Osea mantener los SLAS  a raya.
